```python
from ReusableFunc import load_data
import pandas as pd
import os

# ==========================================
# 1. LOAD DATA
# ==========================================

data = load_data(
    r"C:\Users\Hi\Desktop\Risk-Analysis\data\raw\healthcare_patient_readmission.csv"
)

print("Data loaded successfully!")
print("Original shape:", data.shape)

# ==========================================
# 2. CREATE WORKING COPY
# ==========================================

PrData = data.copy()

# ==========================================
# 3. HANDLE MISSING VALUES
# ==========================================

# Age - median
PrData["Age"] = PrData["Age"].fillna(
    PrData["Age"].median()
)

# Insurance - Unknown
PrData["Insurance_Type"] = PrData["Insurance_Type"].fillna(
    "Unknown"
)

# Average Glucose - diabetes-specific mean
mean_glucose_diabetic = PrData.loc[
    PrData["Diabetes"] == 1,
    "Average_Glucose"
].mean()

mean_glucose_non_diabetic = PrData.loc[
    PrData["Diabetes"] == 0,
    "Average_Glucose"
].mean()

PrData.loc[
    (PrData["Diabetes"] == 1) &
    (PrData["Average_Glucose"].isna()),
    "Average_Glucose"
] = mean_glucose_diabetic

PrData.loc[
    (PrData["Diabetes"] == 0) &
    (PrData["Average_Glucose"].isna()),
    "Average_Glucose"
] = mean_glucose_non_diabetic

# Follow-up
PrData["Followup_Attended"] = PrData[
    "Followup_Attended"
].fillna("Unknown")

# Satisfaction score - median
PrData["Satisfaction_Score"] = PrData[
    "Satisfaction_Score"
].fillna(
    PrData["Satisfaction_Score"].median()
)

# ==========================================
# 4. REMOVE DUPLICATES
# ==========================================

print("Duplicates before:", PrData.duplicated().sum())

PrData = PrData.drop_duplicates()

print("Duplicates after:", PrData.duplicated().sum())

# ==========================================
# 5. STANDARDIZE CATEGORICAL VARIABLES
# ==========================================

PrData["Gender"] = (
    PrData["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)

PrData["Gender"] = PrData["Gender"].replace({
    "male": "Male",
    "female": "Female"
})

PrData["Admission_Type"] = (
    PrData["Admission_Type"]
    .astype(str)
    .str.strip()
    .str.title()
)

# ==========================================
# 6. FINAL DATA QUALITY CHECK
# ==========================================

print("\nFinal missing values:")
print(PrData.isnull().sum())

print(
    "\nTotal missing values:",
    PrData.isnull().sum().sum()
)

print(
    "\nFinal duplicates:",
    PrData.duplicated().sum()
)

print(
    "\nFinal shape:",
    PrData.shape
)

# ==========================================
# 7. SAVE CLEANED DATA
# ==========================================

output_path = (
    r"C:\Users\Hi\Desktop\Risk-Analysis"
    r"\data\processed\ProData.csv"
)

PrData.to_csv(
    output_path,
    index=False
)

print("\nCleaned data saved successfully!")

# ==========================================
# 8. VERIFY SAVED DATA
# ==========================================

check = pd.read_csv(output_path)

print("\nSaved data shape:", check.shape)

print("\nMissing values in saved file:")
print(check.isnull().sum())

print(
    "\nDuplicates in saved file:",
    check.duplicated().sum()
)
```
