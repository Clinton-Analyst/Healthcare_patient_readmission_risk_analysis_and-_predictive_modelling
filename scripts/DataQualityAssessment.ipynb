# checking outliers for numericalvariables;
 

```python
numerical_cols = data.select_dtypes(exclude="str").columns

for col in numerical_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((data[col] < lower) | (data[col] > upper)).sum()

    print(f"{col}: {count} outliers")

    # save the outlier report as csv to reports
    outlier_report = data[(data[col] < lower) | (data[col] > upper)]
    outlier_report.to_csv(f"C:\\Users\\Hi\\Desktop\\Risk-Analysis\\figures\\{col}_outliers.csv", index=False)
```
