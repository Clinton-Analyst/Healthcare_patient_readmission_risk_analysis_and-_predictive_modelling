1. AGE -Replace Missing with median Ages

```python
# Age
PrData["Age"] = PrData["Age"].fillna(
    PrData["Age"].median()
)
```

2. Insurance Type - Fill Missing with Uknown

```python
# Insurance
PrData["Insurance_Type"] = PrData["Insurance_Type"].fillna(
    "Unknown"
)
```

3. Average Glucose - Fill with Mean 

```python
# Average Glucose Level
PrData["Average_Glucose"] = PrData["Average_Glucose"].fillna(
    PrData["Average_Glucose"].mean()
)
```

4. Follow Up - Fill up with Uknown

```python
# Follow-up
PrData["Followup_Attended"] = PrData[
    "Followup_Attended"
].fillna("Unknown")
```

5. Satisfaction Score - Fill with Median

```python
# Satisfaction
PrData["Satisfaction_Score"] = PrData[
    "Satisfaction_Score"
].fillna(
    PrData["Satisfaction_Score"].median()
)
```
