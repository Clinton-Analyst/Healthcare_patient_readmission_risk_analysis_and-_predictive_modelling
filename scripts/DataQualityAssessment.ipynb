# Data Quality Assessment

```python
# loading data, used a reusable function to load the data from the csv file
from ReusableFunc import load_data
import pandas as pd
import os


data = load_data(
    r"C:\Users\Hi\Desktop\Risk-Analysis\data\raw\healthcare_patient_readmission.csv"
)

print("Data loaded successfully!")
print("Original shape:", data.shape)
```

```python
# creating a copy of the data to avoid modifying the original data
PrData=data.copy()
```

# Variables with Missing values:
Age, Insurance_Type, Average_Glucose, Followup_Attended, Satisfaction_Score

# Handling Missing Values

```python
# checking for missing values in the dataset
#missing_values = PrData.isnull().sum()
#print("Missing values before handling:")
#print(missing_values)

#handling missing values for age
# Age - median
PrData["Age"] = PrData["Age"].fillna(
    PrData["Age"].median()
)

# Insurance - Unknown
PrData["Insurance_Type"] = PrData["Insurance_Type"].fillna(
    "Unknown"
)

# Average Glucose Level
PrData["Average_Glucose"] = PrData["Average_Glucose"].fillna(
    PrData["Average_Glucose"].mean()
)

# Follow-up
PrData["Followup_Attended"] = PrData[
    "Followup_Attended"
].fillna("Unknown")

# Satisfaction score - median
PrData["Satisfaction_Score"] = PrData[
    "Satisfaction_Score"
].fillna(
    PrData["Satisfaction_Score"].median()
)

# CHECKING IF THERE ARE ANY MISSING VALUES LEFT
MissingValuesAfter=PrData.isnull().sum()
print("Missing values after handling:")
print(MissingValuesAfter)
```

# Checking Duplicates: 


```python
# Check for duplicates
duplicates = PrData.duplicated().sum()
print("Duplicates found:")
print(duplicates)
```

# standardize gender and admission type

```python
PrData["Gender"] = PrData["Gender"].astype(str).str.strip().str.lower()

PrData["Gender"] = PrData["Gender"].replace({
    "male": "Male",
    "FEMALE": "Female"
   
})
PrData["Gender"].isna().sum()

PrData["Admission_Type"] = PrData["Admission_Type"].astype(str).str.strip().str.lower()

PrData["Admission_Type"] = PrData["Admission_Type"].replace({
    "Emergency ": "Emergency",
    "emergency": "Emergency"
})
PrData["Admission_Type"].isna().sum()
```

# Admission Type is already standardized

```python
# Admission_Type
PrData["Gender"] = (
    PrData["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)

PrData["Gender"] = PrData["Gender"].replace({
    "male": "Male",
    "FEMALE": "Female"
})

PrData["Admission_Type"] = (
    PrData["Admission_Type"]
    .astype(str)
    .str.strip()
    .str.title()
)

PrData["Admission_Type"] = PrData["Admission_Type"].replace({
    "Emergency ": "Emergency",
    "emergency": "Emergency",
    "EMERGENCY": "Emergency"
})
```

# checking outliers for numericalvariables;
 

```python


numerical_cols = PrData.select_dtypes(exclude="str").columns

for col in numerical_cols:
    Q1 = PrData[col].quantile(0.25)
    Q3 = PrData[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((PrData[col] < lower) | (PrData[col] > upper)).sum()

    print(f"{col}: {count} outliers")

    # save the outlier report as csv to reports
    outlier_report = PrData[(PrData[col] < lower) | (PrData[col] > upper)]
    outlier_report.to_csv(f"C:\\Users\\Hi\\Desktop\\Risk-Analysis\\figures\\{col}_outliers.csv", index=False)
```
